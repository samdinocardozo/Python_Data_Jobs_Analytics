```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_US = df[df['job_country'] == 'United States'].copy()
```

```python
df_US['job_posted_month']= df_US['job_posted_date'].dt.strftime('%B')
```

```python
df_US_pivot = df_US.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc='size')
```

```python
df_US_pivot.reset_index(inplace=True)
df_US_pivot['month_no'] = pd.to_datetime(df_US_pivot['job_posted_month'], format='%B').dt.month
df_US_pivot.sort_values('month_no',inplace=True)
df_US_pivot.set_index('job_posted_month', inplace=True)
df_US_pivot.drop(columns='month_no', inplace=True)
```

```python
df_US_pivot
```

```python
top_3 = df_US['job_title_short'].value_counts().head(3).index.tolist()
```

```python
# Create explicit figure and axes
fig, ax = plt.subplots(figsize=(12, 6), dpi=100)

# Plot using a cohesive color palette
df_US_pivot[top_3].plot(
    kind='line', 
    ax=ax, 
    linewidth=2.5, 
    colormap='tab10' # Options: 'viridis', 'Set2', 'coolwarm'
)

# Titles & Labels
ax.set_title('US Pivoted Trends', fontsize=15, fontweight='bold', loc='left', pad=15)
ax.set_xlabel('Months', fontsize=11, fontweight='semibold')
ax.set_ylabel('Count of Jobs', fontsize=11, fontweight='semibold')

# Grid & Spines (Borders)
ax.grid(True, linestyle='--', alpha=0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Legend Outside Frame
ax.legend(
    title='Category', 
    bbox_to_anchor=(1.02, 1), 
    loc='upper left', 
    frameon=False
)

plt.tight_layout()
plt.show()
```
